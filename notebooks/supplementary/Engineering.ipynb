# Decision to use multi-index with redundant levels vs columns
Representing redundant levels, such as experiment parameters, in multiindex instead of columns is at least 3x-5x faster for querying operations and marginally faster for groupby.

```python
import pandas as pd
import numpy as np

# Generate sample data
np.random.seed(42)
n = 100000  # Number of rows
levels1 = np.random.choice(["A", "B", "C", "D"], size=n)
levels2 = np.random.choice(["X", "Y", "Z"], size=n)
levels3 = np.random.choice(["1"], size=n)
levels4 = levels3
values = np.random.randn(n)
data = {
    "level1": levels1,
    "level2": levels2,
    "level3": levels3,
    "level4": levels4,
    "value": values,
}

# Create MultiIndex DataFrame
df_multi = pd.DataFrame(data)
df_multi.set_index(["level1", "level2", "level3", "level4"], inplace=True)
df_multi.sort_index(inplace=True)

# Create Column-Based DataFrame
df_columns = pd.DataFrame(data)
df_columns.set_index(["level1", "level2"], inplace=True)
df_columns.sort_index(inplace=True)


# Define query functions
def query_multi():
    x = df_multi.loc[("A", "X", "1", "1")]
    return x[x["value"] > 0]


def query_columns():
    x = df_columns.loc[("A", "X")]
    x = x[(x["level3"] == "1") & (x["level4"] == "1")]
    return x[x["value"] > 0]


# Benchmark operations
print("Benchmarking Pandas columns vs Pandas multi-index operations with MultiIndex:")

# Filtering
benchmark(query_multi, "Pandas Multi-index Filtering")
benchmark(query_columns, "Pandas Columns Filtering")

# Aggregation
benchmark(
    lambda: df_multi.groupby(["level1", "level2", "level3", "level4"])["value"].mean(),
    "Pandas Multi-index GroupBy Mean",
)
benchmark(
    lambda: df_columns.groupby(["level1", "level2", "level3", "level4"])[
        "value"
    ].mean(),
    "Pandas Columns GroupBy Mean",
)

# Sorting
benchmark(lambda: df_multi.sort_values("value"), "Pandas Multi-index Sorting")
benchmark(lambda: df_columns.sort_values("value"), "Pandas Columns Sorting")

# Join
df_multi2 = df_multi.copy()
df_columns2 = df_columns.copy()
df_multi2.rename(columns={"value": "join_value"}, inplace=True)
df_columns2.rename(columns={"value": "join_value"}, inplace=True)

benchmark(
    lambda: df_multi.merge(df_multi2, left_on="value", right_on="join_value"),
    "Pandas Multi-index Join",
)
benchmark(
    lambda: df_columns.merge(df_columns2, left_on="value", right_on="join_value"),
    "Pandas Columns Join",
)
```

# Decision to prioritize indexing by multi-index key over naive filtering

```python
import pandas as pd
import numpy as np


def filter_df(
    df: pd.DataFrame, conds: dict[str, list] = None
) -> pd.DataFrame:
    """
    Filter a DataFrame according to conditions specified in a dictionary.

    Args:
        df (pd.DataFrame): DataFrame containing experiment data.
        conds (dict[str, list]): Dictionary mapping column or index level names to values to filter on.
        attempt_index (bool): Flag indicating whether to attempt converting conds to multi-index key. Defaults to True.

    Returns:
        pd.DataFrame: Filtered DataFrame.
    """
    if conds:
        mask = np.full(len(df), True)
        for k, v in conds.items():
            v = np.atleast_1d(v)
            if k in df.index.names:  # Filter on index levels
                mask &= df.index.get_level_values(k).isin(v)
            elif k in df.columns:  # Filter on columns
                mask &= df[k].isin(v)
        return df[mask]
    return df


# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [
        ("subject1", 1, "blockA"),
        ("subject1", 1, "blockB"),
        ("subject1", 2, "blockA"),
        ("subject2", 1, "blockA"),
        ("subject2", 2, "blockB"),
        ("subject3", 1, "blockA"),
    ],
    names=["subject", "session", "block"],
)

# Create DataFrame with MultiIndex
data = {
    "metric1": [10, 15, 10, 20, 25, 30],
    "metric2": [5, 10, 15, 20, 25, 30],
}
df = pd.DataFrame(data, index=index)
idx = pd.IndexSlice
conds = {"subject": "subject1", "session": 1, "block": "blockA"}
benchmark(lambda: filter_df(df, conds), "Pandas Multi-index Filtering")
benchmark(lambda: df.loc[tuple(conds.values())], "Pandas Multi-index Key")
benchmark(
    lambda: df.loc[idx[conds["subject"], conds["session"], conds["block"]]],
    "Pandas IndexSlice",
)
```

# Decision to use polars over pandas for huge dataframes

```python
import pandas as pd
import polars as pl
import numpy as np
import timeit

# Generate sample data
np.random.seed(42)
n = 100000  # Number of rows
levels1 = np.random.choice(["A", "B", "C", "D"], size=n)
levels2 = np.random.choice(["X", "Y", "Z"], size=n)
levels3 = np.random.choice(["1"], size=n)
levels4 = levels3
values = np.random.randn(n)
data = {
    "level1": levels1,
    "level2": levels2,
    "level3": levels3,
    "level4": levels4,
    "value": values,
}

# Convert to Pandas and Polars DataFrame
pdf = pd.DataFrame(data)
pdf.set_index(["level1", "level2", "level3", "level4"], inplace=True)
pdf.sort_index(inplace=True)

pldf = (
    pl.DataFrame(data)
    .with_columns(
        pl.col("level1").cast(pl.Categorical),
        pl.col("level2").cast(pl.Categorical),
        pl.col("level3").cast(pl.Categorical),
        pl.col("level4").cast(pl.Categorical),
    )
    .sort(["level1", "level2", "level3", "level4"])
)

# Benchmark operations
print("Benchmarking Pandas vs Polars DataFrame operations with MultiIndex:")

# Filtering
benchmark(lambda: pdf.loc[pdf["value"] > 0], "Pandas Filtering")
benchmark(lambda: pldf.filter(pl.col("value") > 0), "Polars Filtering")

# Aggregation
benchmark(
    lambda: pdf.groupby(["level1", "level2", "level3", "level4"])["value"].mean(),
    "Pandas GroupBy Mean",
)
benchmark(
    lambda: pldf.group_by(["level1", "level2", "level3", "level4"]).agg(
        pl.col("value").mean()
    ),
    "Polars GroupBy Mean",
)

# Sorting
benchmark(lambda: pdf.sort_values("value"), "Pandas Sorting")
benchmark(lambda: pldf.sort("value"), "Polars Sorting")

# Join
pdf2 = pdf.copy()
pldf2 = pldf.clone()
pdf2.rename(columns={"value": "join_value"}, inplace=True)
pldf2 = pldf2.rename({"value": "join_value"})

benchmark(
    lambda: pdf.merge(pdf2, left_on="value", right_on="join_value"), "Pandas Join"
)
benchmark(
    lambda: pldf.join(pldf2, left_on="value", right_on="join_value"), "Polars Join"
)
```

# Decision to use gamma.cdf over matrix exponential
Familiar adage holds: always go built-in over custom whenever you can.

```python
from scipy.linalg import expm
from scipy.stats import gamma
import numpy as np

latent = 5
t = 5
is_avail = 1
n_states = 11
shape = 10
p0 = np.zeros(shape + 1)
p0[0] = 1

# METHOD 1: full state vector
def method1():
    rate = 1 / latent
    Lambda = np.diag(np.ones(n_states) * -rate * (shape), k=0) + np.diag(
        np.ones(shape) * rate * shape, k=1
    )
    Lambda[-1, -1] = 0.0  # no probability of leaving once in the nth state
    p_t = p0 @ expm(Lambda * t)
    p_t = np.clip(p_t, 0.0, 1.0)

    # the last element is the probability of being in the last state ie. reward being available
    if is_avail == 0:  # if reward is not available, do the complementary probability
        return 1.0 - p_t[-1]
    return p_t[-1]


benchmark(method1, "Method 1: Matrix Exponential")
benchmark(lambda: gamma.cdf(t, shape, scale=latent / shape), "Method 2: gamma.cdf")

tol = 1e-10
assert abs(gamma.cdf(t, shape, scale=latent / shape) - method1()) < tol
```

# Verifying Bayesian Update

This shows the posterior converge to the true rate with unbiased, fixed rate sampling of the reward process

We verify by eye that the bayesian update is correct in the following simple setup of inferring the parameter of a Bernoulli random process. In this setup, starting with a uniform prior and always observing a success should result in a posterior where each probability is multiplied by its corresponding latent (also a probability) ie. after the nth update, the posterior probabilities should be proportionate to their corresponding latent to the nth power.
#todo: forego this nonsense and just verify against an explicit conjugat prior calculation ie. beta-bernoulli conjugacy where we update the conjugate parameters by hand and the compare to our brute-force implementation that instead takes a discrete set of possible bernoulli parameters and does vanilla bayesian updating

```python
import numpy as np
from scipy.stats import bernoulli
from matplotlib import pyplot as plt

from foraging.utils.models import (
    ExactBayesianUpdateOnProbabilities,
    GammaLikelihood,
    PossibleSchedules,
    Probabilities,
    RewardObservation,
    Posterior
)


# Initialize prior
support = np.arange(0, 1, 0.1)
prior = Probabilities(support, 1/len(support)*np.ones(len(support)))

# Compose the belief update rule
# likelihood = lambda o, x: bernoulli.pmf(o, x)
update = ExactBayesianUpdateOnProbabilities(likelihood=bernoulli.pmf, vectorize = True)

# Now, compose the posterior
posterior = Posterior(0, prior, update)

# Now, simulate belief updates
true_p = 0.5
n_times = 500
obss = bernoulli.rvs(true_p, size = n_times)
for i, obs in enumerate(obss):
    posterior.update(i, obs)

probs = np.asarray(posterior.representation)
plt.figure(figsize=(10, 20))
plt.imshow(probs[:, :].T, aspect=20)
plt.yticks(
    range(0, len(support), 1),
    ["%.1f" % x for x in support[range(0, len(support), 1)]],
);
```

```python
import numpy as np
from scipy.stats import bernoulli, beta
from matplotlib import pyplot as plt

from foraging.utils.models import (
    ExactBayesianUpdateOnProbabilities,
    Probabilities,
    Posterior
)


# Initialize prior
support = np.arange(0, 1, 0.1)
prior = Probabilities(support, 1/len(support)*np.ones(len(support)))

# Compose the belief update rule
# likelihood = lambda o, x: bernoulli.pmf(o, x)
update = ExactBayesianUpdateOnProbabilities(likelihood=bernoulli.pmf, vectorize = True)

# Now, compose the posterior
posterior = Posterior(0, prior, update)

# Set up conjugate prior equivalent
a, b = 1, 1

# Now, simulate belief updates and compare to conjugate prior probabilities
true_p = 0.5
n_times = 500
obss = bernoulli.rvs(true_p, size = n_times)
for i, obs in enumerate(obss):
    
    # Belief updates
    posterior.update(i, obs)
    a += obs
    b += 1 - obs

    # Compare probabilities
    conjugate_probs = np.zeros_like(support)
    for j, x in enumerate(support):
        conjugate_probs[j] = beta.pdf(x, a = a, b = b)    
    conjugate_probs /= conjugate_probs.sum()
    np.testing.assert_allclose(posterior[i].representation, conjugate_probs)
```

Example on foraging task

```python
import numpy as np
from scipy.stats import gamma
from matplotlib import pyplot as plt

from foraging.utils.models import (
    ExactBayesianUpdateOnProbabilities,
    GammaLikelihood,
    PossibleSchedules,
    Probabilities,
    RewardObservation,
    Posterior
)


# Initialize prior
schedules = PossibleSchedules(schedule=np.arange(1, 30, 0.1), shape=1.0)
prior = Probabilities(schedules, 1/len(schedules)*np.ones(len(schedules)))

# Compose the belief update rule
likelihood = GammaLikelihood()
update = ExactBayesianUpdateOnProbabilities(likelihood=likelihood, vectorize = True)

# Now, compose the posterior
posterior = Posterior(0, prior, update)

# Now, simulate belief updates
true_rate = 5
shape = 10
n_times = 1000
push_interval = true_rate
for i in range(1, n_times + 1):
    rt = gamma.rvs(shape, scale=true_rate / shape)
    ro = push_interval > rt  # this assumes box resets every push
    obs = RewardObservation(is_available=ro, time = push_interval)
    posterior.update(i, obs)

probs = np.asarray(posterior.representation)
plt.figure(figsize=(10, 20))
plt.imshow(probs[:, :].T, aspect="equal")
plt.yticks(
    range(0, len(schedules), 10),
    ["%d" % x for x in schedules[range(0, len(schedules), 10)]],
);
```
